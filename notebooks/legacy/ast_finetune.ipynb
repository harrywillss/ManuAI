```python
# IMPORTS
import os
import json
import random
import numpy as np

import torch
import torch.nn as nn
import evaluate

from transformers import (
    ASTForAudioClassification, 
    ASTFeatureExtractor, 
    ASTConfig,
    TrainingArguments, 
    Trainer, 
    EarlyStoppingCallback,
    set_seed as t_set_seed,
)
from transformers.integrations import TensorBoardCallback
from datasets import Dataset, DatasetDict
from peft import LoraConfig, get_peft_model
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score
from sklearn.utils.class_weight import compute_class_weight

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def set_seed(seed):
    random.seed(seed) 
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    t_set_seed(seed)
```

```python
# Memory optimization
torch.backends.cudnn.benchmark = True  # Faster for fixed input sizes
if torch.backends.mps.is_available():
    # MPS memory growth
    os.environ['PYTORCH_MPS_HIGH_WATERMARK_RATIO'] = '0.0'
print("Memory optimizations applied.")
```

```python
# PARAMETERS
import yaml
from pathlib import Path

with open('ManuAI.config') as f:
    DATA_DIR = Path(yaml.safe_load(f)['data_directory'])

PROCESSED_DIR = DATA_DIR / 'processed'
if not PROCESSED_DIR.exists():
    raise FileNotFoundError(f'Processed data not found at {PROCESSED_DIR}. Run preprocess_data.ipynb first.')

# Model paths
model_output_dir = "./models/finetuned/ast-base-manuai"
adapters_dir = "./models/finetuned/ast-manuai-lora-adapters"
checkpoints_dir = "./models/finetuned/ast-manuai-checkpoints"
models_dir = "./models/base"
model_name = os.path.join(models_dir, "mit-ast")  # Local base model

# Check if continuing from existing fine-tuned model
if os.path.exists(model_output_dir) and os.path.exists(os.path.join(model_output_dir, "model.safetensors")):
    model_name = model_output_dir
    print(f"⚠️  Resuming from existing fine-tuned model: {model_name}")
else:
    print(f"✅ Loading base model from: {model_name}")

# Verify local model exists
if not os.path.exists(os.path.join(model_name, "model.safetensors")):
    raise FileNotFoundError(f"Model not found at {model_name}. Ensure models/base/mit-ast/ contains model files.")

processor = ASTFeatureExtractor.from_pretrained(model_name)

epochs = 30
batch_size = 8               # Reduced from 16 — fits in 16GB VRAM
eval_batch_size = 16         # Slightly larger since no gradients stored
gradient_accum_steps = 4     # Effective batch = 8×4 = 32
learning_rate = 5e-4         # Good for LoRA fine-tuning
warmup_ratio = 0.1
lora_rank = 24
seed = 21

# MPS doesn't support multiprocessing workers or pin_memory
dataloader_num_workers = 0   # Required: collate fns can't be pickled in notebooks
dataloader_pin_memory = False  # MPS doesn't support pinned memory

device = "mps" if torch.backends.mps.is_available() else "cpu"
print(f"Using device: {device}")

set_seed(seed)

accuracy_metric = evaluate.load("accuracy")
precision_metric = evaluate.load("precision")
recall_metric = evaluate.load("recall")
f1_metric = evaluate.load("f1")

print(f"\nData directory: {DATA_DIR}")
print(f"Processed directory: {PROCESSED_DIR}")
print(f"\nTraining config:")
print(f"  Batch size: {batch_size} × {gradient_accum_steps} accum = {batch_size * gradient_accum_steps} effective")
print(f"  Learning rate: {learning_rate}")
print(f"  Epochs: {epochs} (with early stopping)")
```

```python
# Spectrogram augmentations (SpecAugment-style, applied during training)
def spec_augment(spec: torch.Tensor, freq_mask_param: int = 24, time_mask_param: int = 48, n_freq_masks: int = 2, n_time_masks: int = 2) -> torch.Tensor:
    """Apply SpecAugment (frequency and time masking) to a spectrogram tensor."""
    spec = spec.clone()
    _, time_frames, freq_bins = spec.shape if spec.dim() == 3 else (1, *spec.shape)
    
    # Frequency masking
    for _ in range(n_freq_masks):
        f = random.randint(0, min(freq_mask_param, freq_bins - 1))
        f0 = random.randint(0, freq_bins - f)
        if spec.dim() == 3:
            spec[:, :, f0:f0+f] = 0
        else:
            spec[:, f0:f0+f] = 0
    
    # Time masking
    for _ in range(n_time_masks):
        t = random.randint(0, min(time_mask_param, time_frames - 1))
        t0 = random.randint(0, time_frames - t)
        if spec.dim() == 3:
            spec[:, t0:t0+t, :] = 0
        else:
            spec[t0:t0+t, :] = 0
    
    return spec

APPLY_SPECAUGMENT = True  # Toggle spectrogram augmentation during training
print(f"SpecAugment enabled: {APPLY_SPECAUGMENT}")
```

```python
# LOAD PRE-COMPUTED SPECTROGRAMS FROM .npy FILES
# Data is already split into train/val/test folders by preprocess_data.ipynb

from datasets import DatasetDict

def discover_npy_files(split_dir: Path) -> tuple[list[str], list[int]]:
    """Discover all .npy files in a split directory, return paths and labels."""
    paths, lbls = [], []
    for cls_dir in sorted(split_dir.iterdir()):
        if not cls_dir.is_dir():
            continue
        cls_name = cls_dir.name
        if cls_name not in label_to_id:
            print(f"[warn] Unknown class '{cls_name}' — skipping")
            continue
        lbl_id = label_to_id[cls_name]
        for npy_file in cls_dir.glob('*.npy'):
            paths.append(str(npy_file))
            lbls.append(lbl_id)
    return paths, lbls

# Discover labels from train directory
labels = sorted([d.name for d in (PROCESSED_DIR / 'train').iterdir() if d.is_dir()])
label_to_id = {lbl: i for i, lbl in enumerate(labels)}
id_to_label = {i: lbl for lbl, i in label_to_id.items()}
print(f"Classes ({len(labels)}): {labels}")

# Build datasets for each split
def load_split(split_name: str) -> Dataset:
    split_dir = PROCESSED_DIR / split_name
    paths, lbls = discover_npy_files(split_dir)
    return Dataset.from_dict({'path': paths, 'label': lbls})

dataset = DatasetDict({
    'train': load_split('train'),
    'validation': load_split('val'),
    'test': load_split('test'),
})

print(f"Train: {len(dataset['train'])}, Val: {len(dataset['validation'])}, Test: {len(dataset['test'])}")
```

```python
def collate_fn(batch, training=False):
    input_values = []
    labels_out = []

    for item in batch:
        # Load pre-computed spectrogram — already (1024, 128) from preprocess_data.ipynb
        spec = np.load(item['path']).astype(np.float32)  # (time, mel) = (1024, 128)
        spec = torch.from_numpy(spec)
        
        # Verify shape — should be (1024, 128)
        if spec.shape != (1024, 128):
            raise ValueError(f"Unexpected spec shape {spec.shape} from {item['path']}. Expected (1024, 128).")

        # SpecAugment during training
        if training and APPLY_SPECAUGMENT:
            spec = spec_augment(spec)

        input_values.append(spec)
        labels_out.append(item['label'])

    return {
        'input_values': torch.stack(input_values),  # (B, 1024, 128)
        'labels': torch.tensor(labels_out, dtype=torch.long),
    }

def train_collate(batch):
    return collate_fn(batch, training=True)

def eval_collate(batch):
    return collate_fn(batch, training=False)

config = ASTConfig.from_pretrained(
    model_name,
    num_labels=len(labels),
    id2label=id_to_label,
    label2id=label_to_id
)
base_model = ASTForAudioClassification.from_pretrained(
    model_name,
    config=config,
    ignore_mismatched_sizes=True
)

lora = LoraConfig(
    r=lora_rank,
    lora_alpha=lora_rank * 2,
    lora_dropout=0.1,
    bias="none",
    target_modules=["query", "value"],
    modules_to_save=["classifier"]
)

model = get_peft_model(base_model, lora).to(device)
```

```python
# DATA SUMMARY
# Data is already split by preprocess_data.ipynb — no sklearn split needed

for split_name, split_data in dataset.items():
    label_counts = {}
    for lbl in split_data['label']:
        label_counts[id_to_label[lbl]] = label_counts.get(id_to_label[lbl], 0) + 1
    print(f"\n{split_name.upper()} ({len(split_data)} samples):")
    for lbl_name, cnt in sorted(label_counts.items()):
        print(f"  {lbl_name}: {cnt}")
```

```python
# MODEL SETUP
def trainable_parameters(model):
    params, trainable = 0, 0
    for _, p in model.named_parameters():
        params += p.numel()
        trainable += p.numel() if p.requires_grad else 0
    print(f"{model.__class__.__name__} trainable params: {trainable:,}/{params:,} ({100*trainable/params:.2f}%)")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=-1)
    acc = accuracy_metric.compute(predictions=preds, references=labels)["accuracy"]
    prec = precision_metric.compute(predictions=preds, references=labels, average="weighted")["precision"]
    rec = recall_metric.compute(predictions=preds, references=labels, average="weighted")["recall"]
    f1 = f1_metric.compute(predictions=preds, references=labels, average="weighted")["f1"]
    try:
        auc = roc_auc_score(labels, torch.nn.functional.softmax(torch.tensor(logits), dim=-1).numpy(), multi_class='ovr', average='weighted')
    except ValueError:
        auc = 0.0  # Not enough classes in batch
    return {"accuracy": acc, "precision": prec, "recall": rec, "f1": f1, "auc": auc}

class WeightedTrainer(Trainer):
    """Trainer with class-weighted loss and separate collate functions for train/eval."""
    def __init__(self, *args, class_weights=None, train_collate_fn=None, eval_collate_fn=None, **kwargs):
        super().__init__(*args, **kwargs)
        self.class_weights = class_weights
        self._train_collate_fn = train_collate_fn
        self._eval_collate_fn = eval_collate_fn
    
    def get_train_dataloader(self):
        if self._train_collate_fn:
            self.data_collator = self._train_collate_fn
        return super().get_train_dataloader()
    
    def get_eval_dataloader(self, eval_dataset=None):
        if self._eval_collate_fn:
            self.data_collator = self._eval_collate_fn
        return super().get_eval_dataloader(eval_dataset)
    
    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.get("labels")
        outputs = model(**inputs)
        logits = outputs.get("logits")
        loss_fn = nn.CrossEntropyLoss(weight=self.class_weights, label_smoothing=0.1)
        loss = loss_fn(logits.view(-1, self.model.config.num_labels), labels.view(-1))
        return (loss, outputs) if return_outputs else loss

# Training arguments — optimized for M1 Pro 16GB
training_args = TrainingArguments(
    output_dir=checkpoints_dir,
    logging_dir=f"{checkpoints_dir}/runs",
    
    # Learning rate schedule
    learning_rate=learning_rate,
    lr_scheduler_type="cosine_with_restarts",
    warmup_ratio=warmup_ratio,
    weight_decay=0.01,
    
    # Batch settings — memory-optimized for 16GB
    per_device_train_batch_size=batch_size,
    per_device_eval_batch_size=eval_batch_size,
    gradient_accumulation_steps=gradient_accum_steps,
    
    # Training duration
    num_train_epochs=epochs,
    
    # Evaluation & checkpoints
    eval_strategy="steps",
    eval_steps=200,
    save_strategy="steps",
    save_steps=200,
    load_best_model_at_end=True,
    metric_for_best_model="f1",
    greater_is_better=True,
    
    # Logging
    logging_steps=25,
    report_to="tensorboard",
    
    # Memory optimization
    save_total_limit=2,
    dataloader_num_workers=dataloader_num_workers,  # 0 — required for notebook collate fns
    dataloader_pin_memory=dataloader_pin_memory,    # False — MPS doesn't support
    gradient_checkpointing=False,  # Trades compute for memory if needed
    remove_unused_columns=False,   # CRITICAL: Keep 'path' column for collate_fn
    
    # MPS settings (Apple Silicon Devices)
    use_mps_device=(device == "mps"),
    bf16=False,
    fp16=False,  # MPS doesn't support fp16 training well
)

# Callbacks
callbacks = [
    EarlyStoppingCallback(early_stopping_patience=5),
    TensorBoardCallback(),
]

# Load model with LoRA
print(f"Loading model from: {model_name}")
base_model = ASTForAudioClassification.from_pretrained(
    model_name,
    num_labels=len(labels),
    id2label=id_to_label,
    label2id=label_to_id,
    ignore_mismatched_sizes=True,
)

lora_config = LoraConfig(
    r=lora_rank,
    lora_alpha=lora_rank * 2,
    lora_dropout=0.1,
    bias="none",
    target_modules=["query", "value"],
    modules_to_save=["classifier"],
)

peft_model = get_peft_model(base_model, lora_config).to(device)

# Class weights for imbalanced data
train_labels = dataset["train"]["label"]
class_weights = torch.tensor(
    compute_class_weight('balanced', classes=np.unique(train_labels), y=train_labels),
    dtype=torch.float32,
).to(device)

# Trainer
# NOTE: Don't pass processing_class — it strips 'path' column. We handle preprocessing in collate_fn.
trainer = WeightedTrainer(
    model=peft_model,
    args=training_args,
    train_dataset=dataset["train"],
    eval_dataset=dataset["validation"],
    compute_metrics=compute_metrics,
    train_collate_fn=train_collate,
    eval_collate_fn=eval_collate,
    callbacks=callbacks,
    class_weights=class_weights,
)

print(f"\nLabels: {labels}")
trainable_parameters(peft_model)
print(f"Class weights: {dict(zip(labels, [f'{w:.2f}' for w in class_weights.cpu().tolist()]))}")
print("\n✅ Ready to train.")
```

```python
# TRAIN MODEL
result = trainer.train()
trainer.log_metrics("train", result.metrics)
trainer.save_metrics("train", result.metrics)
```

```python
# EVALUATE ON TEST SET 
metrics = trainer.evaluate(eval_dataset=dataset["test"])
trainer.log_metrics("test", metrics)
trainer.save_metrics("test", metrics)
```

```python
# SAVE MODEL
# Merge LoRA weights and save full model
merged_model = peft_model.merge_and_unload()
merged_model.save_pretrained(model_output_dir)
processor.save_pretrained(model_output_dir)
print(f"Full model saved to {model_output_dir}")

# Also save LoRA adapters separately (for future fine-tuning)
peft_model.save_pretrained(adapters_dir)
print(f"LoRA adapters saved to {adapters_dir}")
```

```python
# MODEL METRIC EVALUATION REPORT
print(metrics)

# Per-class
predictions = trainer.predict(dataset["test"])
y_true = predictions.label_ids
y_pred = np.argmax(predictions.predictions, axis=-1)
logits = predictions.predictions  # shape (N, n_classes)
probs = torch.nn.functional.softmax(torch.tensor(logits), dim=-1).numpy()

accuracy_per_class = {}
for i, label_name in enumerate(labels):
    idx = (y_true == i) 
    accuracy_per_class[label_name] = (y_pred[idx] == y_true[idx]).mean()

print("Accuracy per class:")
for label, acc in accuracy_per_class.items():
    print(f"{label}: {acc:.2f}")

print("\nClassification Report:")
report = classification_report(y_true, y_pred, target_names=labels)
print(report)

# Confusion Matrix (plot)
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.savefig("confusion_matrix.png")
plt.show()
```
