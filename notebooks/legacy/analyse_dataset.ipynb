```python
# Count audio files per class in archive/ (raw data from DOC sources)
archive_folder = "archive/"

archive_class_counts = {}
for doc_folder in os.listdir(archive_folder):
    doc_path = os.path.join(archive_folder, doc_folder)
    if not os.path.isdir(doc_path):
        continue
    # Navigate into nested structure (e.g., DOC_001_Tier1/DOC_001_Tier1/train_audio/)
    for root, dirs, files in os.walk(doc_path):
        if 'train_audio' in root:
            for f in files:
                if f.endswith(('.wav', '.mp3', '.flac', '.ogg')):
                    # Extract class from filename or parent folder
                    class_name = os.path.basename(root) if os.path.basename(root) != 'train_audio' else 'unknown'
                    # Try to get class from filename pattern if in train_audio directly
                    if class_name == 'train_audio':
                        # Check if there are class subfolders
                        class_name = f.split('_')[0] if '_' in f else 'unknown'
                    archive_class_counts[class_name] = archive_class_counts.get(class_name, 0) + 1

# Sort and print
print("=== ARCHIVE (Raw Audio) ===")
overall_archive = 0
for class_name, count in sorted(archive_class_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"  {class_name}: {count}")
    overall_archive += count
print(f"\nTotal Archive Audio Files: {overall_archive}")
```

```python
# Count .npy spectrograms per class in data/processed/ (train/val/test splits)
processed_folder = "data/processed/"

split_counts = {}
class_totals = {}

for split in ['train', 'val', 'test']:
    split_path = os.path.join(processed_folder, split)
    if not os.path.isdir(split_path):
        print(f"Split '{split}' not found")
        continue
    
    split_counts[split] = {}
    for class_name in sorted(os.listdir(split_path)):
        class_path = os.path.join(split_path, class_name)
        if not os.path.isdir(class_path):
            continue
        npy_count = len([f for f in os.listdir(class_path) if f.endswith('.npy')])
        split_counts[split][class_name] = npy_count
        class_totals[class_name] = class_totals.get(class_name, 0) + npy_count

# Print per-split counts
print("=== DATA/PROCESSED (Spectrograms) ===\n")
for split, counts in split_counts.items():
    total = sum(counts.values())
    print(f"{split.upper()} ({total} total):")
    for class_name, count in sorted(counts.items()):
        print(f"  {class_name}: {count}")
    print()

# Print totals per class
print("=== TOTALS PER CLASS ===")
overall_total = 0
for class_name, count in sorted(class_totals.items(), key=lambda x: x[1], reverse=True):
    print(f"  {class_name}: {count}")
    overall_total += count
print(f"\nOverall Spectrogram Count: {overall_total}")
print(f"Average per class: {overall_total / len(class_totals):.1f}" if class_totals else "")
```

```python
# Summary comparison: Archive vs Processed
print("=== SUMMARY ===")
print(f"Archive raw audio files: {overall_archive}")
print(f"Processed spectrograms:  {overall_total}")
print(f"Number of classes:       {len(class_totals)}")

# Split distribution
if split_counts:
    train_n = sum(split_counts.get('train', {}).values())
    val_n = sum(split_counts.get('val', {}).values())
    test_n = sum(split_counts.get('test', {}).values())
    print(f"\nSplit distribution:")
    print(f"  Train: {train_n} ({100*train_n/overall_total:.1f}%)")
    print(f"  Val:   {val_n} ({100*val_n/overall_total:.1f}%)")
    print(f"  Test:  {test_n} ({100*test_n/overall_total:.1f}%)")
```
